```python
import fitz
import pytesseract
from PIL import Image
import io
pdf_path = "C:\\Users\\yogeshkannah\\Music\\AI\\surya_python_profile_4+.pdf"
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""
    for page in doc:
        text += page.get_text()
    return text  

extracted_text = extract_text_from_pdf(pdf_path)      
extracted_text             
```
